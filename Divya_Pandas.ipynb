### Q1. NumPy Broadcasting

Two NumPy arrays have shapes `(3, 1)` and `(1, 3)`. Predict the output and explain broadcasting.

```python
a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])

result = a + b
print(result)
print(result.shape)

result_two_values = a + np.array([1, 2])
print(result_two_values)
print(result_two_values.shape)
```

**Explanation:** Array `a` has shape `(3, 1)` and array `b` has shape `(1, 3)`. NumPy broadcasts them to `(3, 3)` and adds the values element-wise. If `b` is replaced with `np.array([1, 2])`, its shape is `(2,)`, which is treated like `(1, 2)` during broadcasting. It is compatible with `(3, 1)`, so the result has shape `(3, 2)`, not an error.

### Q2. CSV Parsing, Missing Values, and Date Formats

Explain how quoted commas are handled, identify possible issues, clean missing marks, and convert dates safely.

```python
csv_data = '''ID,Name,Date,Marks
501,"Mira, K",2026-10-05,74
502,Arnav,05/10/2026,
503,Kiara,2026-10-07,NA'''

df = pd.read_csv(
    StringIO(csv_data),
    na_values=['NA', ''],
    keep_default_na=True
)

def parse_mixed_date(value):
    value = str(value).strip()
    if '-' in value:
        return pd.to_datetime(value, format='%Y-%m-%d', errors='coerce')
    return pd.to_datetime(value, dayfirst=True, errors='coerce')

df['Date'] = df['Date'].apply(parse_mixed_date)
df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')

print(df)
print("\nData types:")
print(df.dtypes)
```

**Explanation:** The comma inside `"Mira, K"` is handled correctly because the value is enclosed in quotes. The Date column has mixed formats, so ISO dates and slash dates are parsed carefully. The Marks column contains blank and `NA` values, which become missing values. The value `NA` should be investigated before cleaning because it may mean missing, not applicable, or not assessed depending on the dataset.

### Q3. Extreme Values in Salary Data

Predict the mean and median, explain the impact of the extreme value, choose a better typical statistic, and identify the highest salary record.

```python
df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D"],
    "Salary": [25000, 30000, 35000, 10000000]
})

print("Mean:", df["Salary"].mean())
print("Median:", df["Salary"].median())

highest_index = df["Salary"].idxmax()
print("\nHighest salary record:")
print(df.loc[highest_index])
```

**Explanation:** The salary `10000000` pulls the mean upward, so the mean no longer represents a typical employee's salary well. The median is better here because it is less affected by the extreme value. The highest salary record should be investigated before removal because it may be a data-entry error or a genuine special case.

### Student Assessment Data Cleaning

The dataset is created inside the notebook so the GitHub folder can contain only the notebook and README file.

```python
csv_data = '''Student_ID,Student_Name,Department,Assessment,Marks
D201,Ishika,IT,Python,83
D202,Manav,it,Python,76
D203,Kavya,I.T.,Python,94
D204,Rohan,HR,Python,
D205,Tanvi,HR,Python,Absent
D206,Nikhil,Finance,Python,5
D207,Ananya,finance,Python,69
D208,Vihaan,IT,Python,72
D202,Manav,it,Python,76
D209,Priyanshi,I.T.,Python,NA
D210,Samar,HR,Python,86 marks
D211,Meenal,Finance,Python,89
D212,Devika,IT,Python,
D213,Aditya,HR,Python,61
D214,Ritika,Finance,Python,112'''

df = pd.read_csv(StringIO(csv_data), keep_default_na=True)

print("First 5 rows:")
print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())
```

**Explanation:** `head()` displays the first five records and `isnull().sum()` counts missing values in each column. The Marks column has missing entries and mixed text values.

```python
df['Department'] = (
    df['Department']
    .astype('string')
    .str.strip()
    .str.upper()
    .replace({
        'I.T.': 'IT',
        'FINANCE': 'Finance'
    })
)

print(df['Department'].unique())
```

**Explanation:** Department names are standardized by removing extra spaces, converting text to uppercase, and replacing known variations such as `I.T.` and `finance` with consistent names.

```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate records:", duplicate_count)

df = df.drop_duplicates()
print("Number of records after removing duplicates:", len(df))
```

**Explanation:** `duplicated()` identifies exact duplicate rows, and `drop_duplicates()` removes them. One duplicate student assessment record is removed.

```python
df['Original_Marks'] = df['Marks']

df['Marks'] = (
    df['Marks']
    .astype('string')
    .str.replace('marks', '', case=False, regex=False)
    .str.strip()
)

df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')
print(df[['Original_Marks', 'Marks']])
```

**Explanation:** The text `86 marks` is cleaned to `86`. Non-numeric values such as `Absent`, `NA`, and blank entries become missing values, while valid numeric marks remain numeric.

```python
absent_students = df[
    df['Original_Marks'].astype('string').str.strip().str.lower() == 'absent'
]

missing_marks = df[df['Marks'].isna()]

print("Absent Students:")
print(absent_students[['Student_ID', 'Student_Name', 'Original_Marks']])

print("\nStudents with Missing Marks:")
print(missing_marks[['Student_ID', 'Student_Name', 'Original_Marks']])
```

**Explanation:** Students marked as `Absent` should be kept as a separate case instead of being treated the same as marks that were not entered. Deleting missing values without investigation may remove useful records and create biased analysis.

```python
invalid_marks = df[(df['Marks'] < 0) | (df['Marks'] > 100)]

print("Invalid Marks:")
print(invalid_marks[['Student_ID', 'Student_Name', 'Marks']])

df.loc[(df['Marks'] < 0) | (df['Marks'] > 100), 'Marks'] = pd.NA
```

**Explanation:** Marks should usually be between 0 and 100, so `112` is invalid. It should be investigated before deletion because it may be a data-entry error or a special grading case. Until verified, it is safer to treat it as missing.

```python
print("Cleaned DataFrame:")
print(df)

print("\nRecords removed:", duplicate_count)
print("Missing marks remaining:", df['Marks'].isna().sum())
print("Invalid marks identified:", len(invalid_marks))
```

**Final Explanation:** The dataset is cleaned by standardizing department names, removing exact duplicate records, converting marks to numeric values, separating absent students from missing entries, and identifying invalid marks outside the 0-100 range.
